```python
# @title Khôi phục snapshot nguồn đã ghim
SOURCE_MANIFEST = json.loads('{"base_git_revision": "af282252ed56190a15d12c17eb258f3bc875f844", "archive_sha256": "21545d22b4ea5816d07775e663414472ceef0ae7cb2bce4ecde11f62195f60f5", "files_sha256": {".env.example": "4e1e224132f5eec4cb900325f98d24707b5ec70da9d52c25a66346d16c553732", "app.py": "51231a22afd0830e612adf2c5bc8e7ec8600208961c90ce5600a7dde21f9fcab", "benchmarks/check_demo.py": "9edd62612069f5c3cb114ceb9970a8820830756488492711ca35c01d4096f71a", "benchmarks/colab-constraints.txt": "229dfc11d95c58607502317d58ab1cc60bbed2a91b2bdba83fd26f3188b0062b", "benchmarks/colab_runtime.py": "8d3b6394d3fe51314da384c76a79a78965f5e773ddb34790298aff40bd029369", "benchmarks/evaluate_generation.py": "32bf9bee7bd5579b7e2cbae245e74084f989bd51bdd131734f21a0c5c6e14dff", "benchmarks/evaluate_unicode_robustness.py": "260f333f2483e6389feeafba2cd62deedf273c21e7a9f0b3422ef3a620510418", "benchmarks/passage_reranker.py": "bd31b80731d627c837e91c9c28c35e1663b5627d5559bf376f5d5f58c0713274", "benchmarks/retrieval_study.py": "1fecbd8728f0af8c5cba8cc55eddad38c5d5b6977f8761c90fbc3c448fea948a", "requirements.txt": "6ab285e4a5a1b12ebca9709fb111bdceb31c4f7208d20db60950754ffad3c904", "results/colab/retrieval/nodiac50.json": "e30d7cb7b5224d4c22f8e35db766ebdc92c3409af9461f956f7a972e280052d8", "results/colab/retrieval/nodiac50_rerank.json": "4f10bc4122fbf8e1e419cb2245921042c8e15a8d07a424255949a7bed0d475cd", "results/colab/retrieval/protocol.md": "63b4bf03bfef2069c57fdc6f2e5945e7c60032d23e6cada553057121afe67753", "results/colab/retrieval/selection.json": "8b6d58657385b24ab33040aa331a2cb657c0f9c0bba0c584604f338f2e975f8b", "results/colab/retrieval/sets.json": "04f94354cc188f190244daa056a64a7eba165050336535b24d2b77e51ffcac4d", "results/colab/retrieval/train200.json": "9d6882b713ce49f667127c91bcabfc8012bde3e94f75a17d21f55dbdf8ac59ae", "results/colab/retrieval/train200_rerank.json": "d765b5813cddb917bd8a75ca5cccbd8bd391ae8aa30ab6f790dd690772701506", "results/upgrade_20261123/GENERATION.md": "bf59db6c9f8208ff2bef2b1dc64135861b41e4e4afabef60d6fb6bfd08c58c71", "results/upgrade_20261123/PROTOCOL.md": "bd52c1d24a9913a2d415a27b0ad7edf459681cdb5eceeb7a68aa90a6430994e9", "results/upgrade_20261123/REPORT.md": "8b11dba63be1fc220da478a8ce7bb083ebf2ddef960b2a29da416ad42a2cd3e1", "results/upgrade_20261123/RETRIEVAL.md": "649aad84dcd34506136cf902125cce471065e1bb56d155ba945ea01eb23d92ff", "results/upgrade_20261123/demo.json": "f21b0cbb0330fff6b55d6cfd6188fad9e516b9ff493a2a897b081d1c6519ebd0", "results/upgrade_20261123/demo_final.json": "2243d2c9bd574bc64bea0749be458c2bd542f78b28314a10e02f229057012c19", "results/upgrade_20261123/execution.json": "7faefb0bc94ff53ef53bba3a589552c09a435f098729fb003b3f6c0c73669d8c", "results/upgrade_20261123/final_execution.json": "39eaa1d6d1afea461da7b0e39bee24626833fe37775fef3ac063e94435dcffeb", "results/upgrade_20261123/generation_after.csv": "499c22b0a91552217a5d169680e5144f75fa85e82a0f8adb1874f17d2ecc91bc", "results/upgrade_20261123/generation_after.json": "7ab728ab2408d9a1369d086d36c95093fda1ad99aa7a897fbac6de535185a7f1", "results/upgrade_20261123/generation_before.csv": "6211d3fca6f61f12f68b49d5976b11da4f6dd0e8d40778791662b8c2add442af", "results/upgrade_20261123/generation_before.json": "3bccd7d93ec00f16161a047ad426e400f5ec3cb5c33acb6418f13e9cda7029bc", "results/upgrade_20261123/generation_before180.csv": "ad2a65b6e45ce04f3ca1ee0031e83e4f69cdc08c0235325e2465680c0df1b937", "results/upgrade_20261123/generation_before180.json": "6473e5e29d3fe4d040691658489ecb4816d462784c1215512542d7f1d571db3c", "results/upgrade_20261123/generation_final.csv": "d5427a92646f96c499a6b79829e1c7ecd073d3b70acdc40655bd3a6ef303e03c", "results/upgrade_20261123/generation_final.json": "f00397ad6d632b3c6868b36b936ef90228480647cd296e671c41e8964bd48b82", "results/upgrade_20261123/retrieval/confirmation100.json": "ed1d4039ae42af540f0e2aa7d5de7f83a234f9b6144fbcc4b8ab648fab0a0140", "results/upgrade_20261123/retrieval/confirmation100.jsonl": "7b31cb9e990a87862f7578af2236447c9d7d06f855d67855f1aa20ff17ba6060", "results/upgrade_20261123/retrieval/dev100.json": "93ca2e365801e18f62123a8595cf29be3ad5b0a9d4672a7893680034db29128a", "results/upgrade_20261123/retrieval/dev100.jsonl": "a85bcefab563cb624a5179b8fd56ea76371d67cb4040e9a34029b966a6741e95", "results/upgrade_20261123/robustness/after.json": "c35188979efa8b8d34aed8fd6c6009c7454b8c0f9d4077fa1b59aa5ee508ee6c", "results/upgrade_20261123/robustness/before.json": "4d71bb06d30317c0e09213c88bde291f16a73cdb8fa49fda74fe650b899f56c3", "results/upgrade_20261123/robustness/execution.json": "83d897d7cf80c81337c0cf9ce42a81cb89542e04ad71c144e4a60eb7727112e6", "src/__init__.py": "2fdcbff592b82ac780b7cc2a0cdb771ef368d98555cbaab6ad3b60cf32719669", "src/config.py": "3f7fc2036e7e1181c6aa93b4dc84f0b1201b68000ef082ab4a114166b0931f7a", "src/dataset.py": "071db5fb6de8381321784ccc4dc805861f1714441ae561b3d6faac70239e9729", "src/embeddings.py": "a5e0e6cd2ff63ae4f9a154ffde39f8b1df53425b722a33ec62e0b1db953a3eab", "src/evaluation.py": "8ee37062b8596cd92193cf44c1179b25def1384093d7b5b869c23961274eb10c", "src/exact_search.py": "570391358606b4724acf6efbf63e0d7119e574f4a684da5946e99c4ac5d45fbe", "src/generation.py": "2514269ab52f180a22a5d67484bf9f2d5a2e01c0958fb7d6563607718c9d4422", "src/ollama_runtime.py": "02315870fb0ace879c83a04c0f1b79adf5cd2d9ee4d4ad9e228a28e175f79861", "src/retrieval.py": "f522b29322d235ff50141dd41ae31fac2443c3a9604ffbda44f8622998d6b132", "src/safety.py": "d94c03248a923b4231e47bdfa79f89f335fe499e98c1922a89498fc585ae135d", "tests/test_evaluation.py": "5d6e1e9402934a89989a9c2f63b0ceed05aad0b599c0a8c50aa9fbfaa764dae9", "tests/test_generation_benchmark.py": "2f122f8467c68690ecb51c063d50bbc6d4f281c50cc3d24cc52fe43394b8125c", "tests/test_generation_spans.py": "c980f661333f55c7dba7f59c5926b67492761c30e91120740708b5df49c9bcac", "tests/test_ollama_runtime.py": "01682f4f4d034ca6c46e1d73b6e9da680484a311a6006b240273d7d8d3a333aa", "tests/test_refusal_pipeline.py": "87da062f7d59e94c5ca7f7a03bd13471d6bfaf034caf2d9c7b1f0950f290072c", "tests/test_retrieval.py": "6086a582c3c928fa3151be3d83cfe38425e30fbdeb2cb80edcf074e30074022a", "tests/test_safety.py": "5b262330756179436075d127bab7ad0f4e78c43075c36dd5bcb279284aae6a44", "tests/test_startup_repair.py": "7147af6f981c2c328d5af8719dffb70bd69c416f8643ff263bc8ee7cc512fbb5"}, "files_changed_since_base": ["benchmarks/check_demo.py", "benchmarks/colab-constraints.txt", "benchmarks/evaluate_generation.py", "benchmarks/evaluate_unicode_robustness.py", "results/upgrade_20261123/GENERATION.md", "results/upgrade_20261123/PROTOCOL.md", "results/upgrade_20261123/REPORT.md", "results/upgrade_20261123/RETRIEVAL.md", "results/upgrade_20261123/demo.json", "results/upgrade_20261123/demo_final.json", "results/upgrade_20261123/execution.json", "results/upgrade_20261123/final_execution.json", "results/upgrade_20261123/generation_after.csv", "results/upgrade_20261123/generation_after.json", "results/upgrade_20261123/generation_before.csv", "results/upgrade_20261123/generation_before.json", "results/upgrade_20261123/generation_before180.csv", "results/upgrade_20261123/generation_before180.json", "results/upgrade_20261123/generation_final.csv", "results/upgrade_20261123/generation_final.json", "results/upgrade_20261123/retrieval/confirmation100.json", "results/upgrade_20261123/retrieval/confirmation100.jsonl", "results/upgrade_20261123/retrieval/dev100.json", "results/upgrade_20261123/retrieval/dev100.jsonl", "results/upgrade_20261123/robustness/after.json", "results/upgrade_20261123/robustness/before.json", "results/upgrade_20261123/robustness/execution.json", "src/generation.py", "src/retrieval.py", "tests/test_generation_benchmark.py", "tests/test_generation_spans.py", "tests/test_retrieval.py", "tests/test_startup_repair.py"], "description": "Exact files embedded in this notebook; the snapshot hash identifies its code."}')
SOURCE_ZIP_BASE64 = ''

def extract_verified_source(encoded, manifest, directory):
    raw = base64.b64decode(encoded, validate=True)
    if hashlib.sha256(raw).hexdigest() != manifest["archive_sha256"]:
        raise RuntimeError("Gói nguồn không khớp SHA-256.")
    directory = Path(directory)
    if directory.is_symlink():
        raise RuntimeError("Thư mục nguồn không được là symlink.")
    directory.mkdir(parents=True, exist_ok=True)
    expected = manifest["files_sha256"]
    with zipfile.ZipFile(io.BytesIO(raw)) as archive:
        if len(archive.namelist()) != len(expected) or set(archive.namelist()) != set(expected):
            raise RuntimeError("Danh sách file trong gói nguồn không khớp manifest.")
        for name, digest in expected.items():
            relative = PurePosixPath(name)
            if relative.is_absolute() or ".." in relative.parts:
                raise RuntimeError("Đường dẫn nguồn không hợp lệ.")
            content = archive.read(name)
            if hashlib.sha256(content).hexdigest() != digest:
                raise RuntimeError("Sai hash của file nguồn: " + name)
            destination = directory.joinpath(*relative.parts)
            if any(part.is_symlink() for part in [destination, *destination.parents]):
                raise RuntimeError("Không ghi qua symlink: " + name)
            if destination.exists():
                if destination.read_bytes() != content:
                    raise RuntimeError("File nguồn đã bị sửa trong runtime: " + name +
                                       ". Dùng thư mục runtime mới để tránh ghi đè.")
            else:
                destination.parent.mkdir(parents=True, exist_ok=True)
                destination.write_bytes(content)
    return directory

CODE_ROOT = Path("/content") / ("vicare-source-" + SOURCE_MANIFEST["archive_sha256"][:12])
extract_verified_source(SOURCE_ZIP_BASE64, SOURCE_MANIFEST, CODE_ROOT)
SESSION["source"] = SOURCE_MANIFEST
(RUN_DIR / "source_manifest.json").write_text(
    json.dumps(SOURCE_MANIFEST, ensure_ascii=False, indent=2), encoding="utf-8")
print("Nguồn:", CODE_ROOT)
print("Git làm nền:", SOURCE_MANIFEST["base_git_revision"])
print("SHA-256 snapshot:", SOURCE_MANIFEST["archive_sha256"])
print("File khác mốc Git:", SOURCE_MANIFEST["files_changed_since_base"] or "không có")
```

**Recall@5** đo việc tài liệu chuẩn có nằm trong 5 kết quả đầu. **Coverage** đo tài liệu
chuẩn có trong index. Hai chỉ số này không phải độ chính xác của câu trả lời.
Dev100 đã được dùng phân tích lỗi; không gọi đây là tập chưa từng xem.


```python
# @title Đo Dense, Hybrid v2, Hybrid HNSW và cấu hình đã chọn
RETRIEVAL_DIR = RUN_DIR / "retrieval"
for suffix in ("json", "jsonl"):
    (RETRIEVAL_DIR / ("dev100." + suffix)).unlink(missing_ok=True)
run_project("retrieval_dev100", "-m", "benchmarks.retrieval_study", "final",
            "--output", RETRIEVAL_DIR, timeout=3600)
dev_report = json.loads((RETRIEVAL_DIR / "dev100.json").read_text(encoding="utf-8"))

def show_retrieval(report):
    rows = ["| Phương pháp | N | Recall@1 | Recall@3 | Recall@5 | Recall@10 | Coverage | p95 (ms) |",
            "|---|---:|---:|---:|---:|---:|---:|---:|"]
    for method, metrics in report["metrics"].items():
        if metrics["n"] != report["metadata"]["expected_count"]:
            raise RuntimeError("Mẫu số đánh giá không đầy đủ.")
        values = [method, str(metrics["n"])]
        values += [f'{metrics["recall"]["@" + str(k)]:.1%}' for k in (1, 3, 5, 10)]
        values += [f'{metrics["coverage"]:.1%}', f'{metrics["latency_ms"]["p95"]:.2f}']
        rows.append("| " + " | ".join(values) + " |")
    display(Markdown("\n".join(rows)))

show_retrieval(dev_report)
delta = 100 * (dev_report["metrics"]["selected"]["recall"]["@5"] -
               dev_report["metrics"]["dense"]["recall"]["@5"])
print(f"Chênh lệch selected − Dense tại Recall@5: {delta:+.1f} điểm phần trăm.")
```

Mọi câu, lỗi hoặc điểm không hợp lệ đều được giữ trong báo cáo. Các file xuất công khai
chứa ID và chỉ số; notebook không xuất toàn bộ câu hỏi/đáp án của bộ dữ liệu.


```python
# @title Đo generation thực tế và hiển thị rõ các trường hợp từ chối
GENERATION_FILE = RUN_DIR / "generation" / "final.json"
if RUN_GENERATION_EVAL:
    GENERATION_FILE.unlink(missing_ok=True)
    GENERATION_FILE.with_suffix(".csv").unlink(missing_ok=True)
    run_project("generation_dev10", "-m", "benchmarks.evaluate_generation", "--method", "Hybrid",
                "--run-label", RUN_ID, "--output", GENERATION_FILE, timeout=3600)
    generation = json.loads(GENERATION_FILE.read_text(encoding="utf-8"))
    if generation["n_questions"] != 10 or len(generation["rows"]) != 10:
        raise RuntimeError("Phép đo không đủ đúng 10 câu đã quy định.")
    rows = generation["rows"]
    answered = sum(row["status"] == "answered" for row in rows)
    positive_answers = sum(row["status"] == "answered" and row["judge_score"] == 1 for row in rows)
    positive_refusals = sum(row["status"] == "refused" and row["judge_score"] == 1 for row in rows)
    fields = [
        ("Judge@10 theo protocol lưu trong báo cáo", f'{generation["judge_hits"]}/10'),
        ("Số câu được chatbot trả lời", f"{answered}/10"),
        ("Câu trả lời được judge chấm 1 / toàn bộ 10 câu", f"{positive_answers}/10"),
        ("Từ chối", f'{generation["refusals"]}/10'),
        ("Từ chối nhưng judge vẫn chấm 1", str(positive_refusals)),
        ("Nhánh cấp cứu", str(generation["status_counts"].get("emergency", 0))),
        ("Judge sai định dạng", str(generation["invalid_judge_format"])),
        ("Lỗi gọi judge", str(generation["judge_request_errors"])),
        ("Lỗi kỹ thuật của ứng dụng", str(generation["technical_failures"])),
    ]
    table = ["| Chỉ số của phiên này | Kết quả |", "|---|---:|"]
    table += [f"| {name} | {value} |" for name, value in fields]
    display(Markdown("\n".join(table)))
    detail = ["| ID | Trạng thái | Mã lý do | Judge |", "|---|---|---|---:|"]
    detail += [f'| {r["id"]} | {r["status"]} | {r["reason_code"]} | {r["judge_score"]} |' for r in rows]
    display(Markdown("\n".join(detail)))
    if positive_refusals:
        print("Lưu ý: judge đã chấm 1 cho câu từ chối. Không gọi Judge@10 là độ chính xác chatbot.")
    if generation["technical_failures"] or generation["judge_request_errors"]:
        SESSION["stages"]["generation_dev10"]["status"] = "COMPLETED_WITH_ERRORS"
        save_session()
        print("Phép đo đã giữ đủ mẫu số nhưng có lỗi kỹ thuật; xem từng dòng và kiểm tra dịch vụ.")
else:
    SESSION["stages"]["generation_dev10"] = {"status": "NOT_RUN"}
    save_session()
    print("Chưa đo generation trong phiên này; không lấy số của phiên khác thay vào.")
```

chứng, ngữ cảnh tăng huyết áp, lỗi dịch vụ và cấp cứu. Lỗi dịch vụ được cô lập trên một

## 8. Nhập câu hỏi và kiểm tra nguồn

Sửa ô câu hỏi bên dưới rồi chạy lại cell. Lựa chọn Dense/Hybrid gọi cùng API `Assistant.ask`
của ứng dụng. Mỗi lượt chạy là một tiến trình riêng để không cài Torch vào kernel Colab;
thời gian hiển thị vì vậy có thêm chi phí nạp chỉ mục. Các phép đo ở trên đã tách riêng
chi phí này theo protocol của dự án.

Tài liệu truy xuất là dữ liệu chưa được thẩm định. Chỉ bật xem nguyên đoạn khi cần đối chiếu;
không công khai output chứa thông tin cá nhân hoặc đoạn chưa rà soát. Gói kết quả cuối
không chứa bản ghi câu hỏi tương tác hay các đoạn nguồn ở cell này.


```python
# @title Hỏi ViCare
QUESTION = "Những yếu tố nào ảnh hưởng đến kết quả đo lưu huyết não?"  # @param {type:"string"}
METHOD = "Hybrid"  # @param ["Dense", "Hybrid"]
SHOW_FULL_SOURCES = False  # @param {type:"boolean"}

if METHOD not in {"Dense", "Hybrid"}:
    raise ValueError("Phương pháp phải là Dense hoặc Hybrid.")
interactive_path = RUNTIME_ROOT / "private-interactive.json"
bridge = r"""
import json, sys
from pathlib import Path
from src.config import Settings
from src.generation import Assistant
from src.safety import validate_url
request = json.load(sys.stdin)
reply = Assistant(Settings.from_env()).ask(request["question"], request["method"])
result = {"answer": reply.answer, "status": reply.status, "reason_code": reply.reason_code,
          "latency_ms": reply.latency_ms, "citation_ids": reply.citation_ids,
          "sources": [{"number": i, "doc_id": hit.chunk.doc_id,
                       "url": validate_url(hit.chunk.url), "text": hit.chunk.text,
                       "cited": str(i) in reply.citation_ids}
                      for i, hit in enumerate(reply.hits, 1)]}
Path(sys.argv[1]).write_text(json.dumps(result, ensure_ascii=False), encoding="utf-8")
"""
import textwrap
run_project("interactive_question", "-c", textwrap.dedent(bridge), interactive_path,
            input_text=json.dumps({"question": QUESTION, "method": METHOD}, ensure_ascii=False),
            timeout=600)
answer = json.loads(interactive_path.read_text(encoding="utf-8"))
display(HTML('<div style="padding:16px;border:1px solid #b9ded9;border-radius:10px;'
             'white-space:pre-wrap">' + html.escape(answer["answer"]) + '</div>'))
print("Trạng thái:", answer["status"], "·", answer["reason_code"],
      "· thời gian xử lý:", answer["latency_ms"], "ms")
for source in answer["sources"]:
    label = f'[{source["number"]}] {source["doc_id"]}' + (" — đã trích dẫn" if source["cited"] else "")
    target = ('<a href="' + html.escape(source["url"], quote=True) +
              '" target="_blank" rel="noopener noreferrer">Mở bài nguồn</a>') if source["url"] else ""
    excerpt = ('<p style="white-space:pre-wrap">' + html.escape(source["text"]) + '</p>') if SHOW_FULL_SOURCES else ""
    display(HTML('<details><summary>' + html.escape(label) + '</summary>' + target + excerpt + '</details>'))
if not answer["sources"]:
    print("Không có thẻ nguồn cho lượt này.")
```
